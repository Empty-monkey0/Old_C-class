# 1. Linked List So Far

### Last class we built a `LinkedList` class managing nodes on the heap. We wrote `insert`, `display`, and an iterative destructor. Here's the full picture, with those three methods filled in.
<br>

```c
////linkedList_recap.cpp
// Example 1a - Linked List from last class (insert, display, iterative destructor filled in)
#include <iostream>

struct Node {
    int data;
    Node* next;
};

class LinkedList {
private:
    Node* head;

public:
    LinkedList() : head(nullptr) {}

    // Iterative destructor (from last class) - save BEFORE delete!
    ~LinkedList(){
        Node* current = head;
        while (current != nullptr){
            Node* nextNode = current->next;   // save BEFORE deleting
            std::cout << "Now deleting node with data " << current->data << "\n";
            delete current;
            current = nextNode;               // advance using the saved next
        }
    }

    void insert(int n){
        if (head != nullptr){
            Node* current = head;
            while (current->next != nullptr){
                current = current->next;
            }
            Node* last = new Node;
            last->data = n;
            last->next = nullptr;
            current->next = last;
        }
        else {
            head = new Node;
            head->data = n;
            head->next = nullptr;
        }
    }

    void display() const {
        Node* current = head;
        while (current != nullptr){
            std::cout << current->data << "->";
            current = current->next;
        }
        std::cout << "nullptr\n";
    }
};

int main() {
    LinkedList list;
    list.insert(10);
    list.insert(20);
    list.insert(30);
    list.insert(40);
    std::cout << "Linked List: ";
    list.display();
    return 0;
}
```

### The four cases to handle:
### 1. **Empty list** — `head == nullptr`. Nothing to do.
### 2. **Target is the head** — no previous node to rewire; move `head` forward, then delete the old head.
### 3. **Target is in the middle or at the tail** — walk until we find it; rewire the previous node's `next`; delete.
### 4. **Target not in the list** — walker falls off the end; nothing to delete.

# 3. Recursive Destructor

### One question that came up: could we do the destructor *recursively* instead of with a `while` loop?

### The idea: write a helper function `deleteAll(Node* n)` that deletes node `n` plus everything reachable after it. Then the destructor becomes one line: `deleteAll(head);`.

### Two things every recursive function needs:
### - a **base case** (when do we stop?)
### - a **recursive case** (how do we make progress toward the base case?)

### **You try!** (Example 3a) Fill in `deleteAll` below.
<br>

```c
////linkedList_recursive_dtor.cpp
// Example 3a (You try!) - recursive destructor helper
#include <iostream>

struct Node { int data; Node* next; };

class LinkedList {
private:
    Node* head;

    // Recursively deletes node n and everything after it
    void deleteAll(Node* n){
        if(n == nullptr){ 
            return; 
            }
        deleteAll(n->next);
        delete n;


    }

public:
    LinkedList() : head(nullptr) {}

    ~LinkedList(){
        deleteAll(head);
    }

    void insert(int n){
        if (head != nullptr){
            Node* current = head;
            while (current->next != nullptr) current = current->next;
            Node* last = new Node; last->data = n; last->next = nullptr;
            current->next = last;
        } else {
            head = new Node; head->data = n; head->next = nullptr;
        }
    }
    void display() const {
        Node* current = head;
        while (current != nullptr){ std::cout << current->data << "->"; current = current->next; }
        std::cout << "nullptr\n";
    }
};

int main(){
    LinkedList list;
    list.insert(10); list.insert(20); list.insert(30);
    list.display();
    return 0;
}
```

### What did your `operator[]` actually return? An `int`. Where does that `int` live? It's a **copy** of the node's `data` — a temporary value with no permanent home, alive just long enough for the surrounding expression. You can't write a value into a copy that's about to disappear.

### To make `list[2] = 5;` work, we need to return something that points back at the **real data** in the node — that's what `int&` (a reference) does. Same lesson as last class's broken `runner = newNode` warmup: writing TO a local copy on the stack vs. writing THROUGH something to the real data on the heap. Reference returns are the THROUGH option.

### **Templates** let us write the function *once* with a placeholder type, and the compiler stamps out a version per type used:


### Three things to notice about `template <typename T>`:

### 1. **`T` is just a type name.** Use it wherever you'd write `int` or `double`. `T` itself never gets `<>` after it — you wouldn't write `int<>`, so don't write `T<>`.
### 2. **`template <typename T>` is a one-shot introduction.** It applies only to the very next definition. Each function (or class, or struct) that uses a template parameter needs its own `template <typename T>` line on top.
### 3. **Templates are 100% compile-time.** Nothing template-related happens at runtime. By the time the program runs, the template is gone, only the generated versions remain in the binary.
<br>

```c
////maxOfTemplate.cpp
// Example 5b - one templated version replaces all three
#include <iostream>
#include <string>

template <typename T>
T maxOf(T a, T b){
    if (a > b) return a;
    return b;
}

int main(){
    std::cout << maxOf(3, 7) << "\n";
    std::cout << maxOf(3.5, 2.1) << "\n";
    std::cout << maxOf<std::string>("apple", "banana") << "\n";
    std::cout << maxOf(3, 2.1) << "\n";   
    return 0;
}
```

<br>
<br><br><br><br><br>
<br><br><br><br><br>


### **A note on what's actually happening at compile time:** A template is a *recipe*, not a function. When the compiler sees `maxOf(3, 7)`, it deduces `T = int`, generates an `int` version, and emits its machine code into the binary. When it sees `maxOf(3.5, 2.1)`, it generates a `double` version. By the time your program runs, the template is gone — the binary contains the same generated functions you would have written by hand.

### Two consequences worth noting:
### - **If you write a template but never use it, the compiler generates nothing.** No instantiation = no code in the binary.
### - **Each unique `<T>` you use produces a new version.** `LinkedList<int>` and `LinkedList<std::string>` give you two distinct generated classes' worth of code.

### To templatize the class, we add `template <typename T>` above the class — and (the part that's easy to miss) we also add it above any *other* definition that uses `T`. Our `Node` struct uses `T` for its `data`, so it gets its own `template <typename T>` line too.

### Three rules to remember when writing the bodies:
### 1. **`T` is a type name; never write `<>` after it.** `T data;` `T n;` `T operator[](int i);` — all of these treat `T` like you would `int`.

When *using* it as a type — including inside the class itself — we write `LinkedList<T>` or `Node<T>`. The constructor's name is still just `LinkedList`.

# 7. `std::list` and Intro to Iterators

### The C++ Standard Template Library (STL) has its own linked list class: `std::list`. Internally it's a doubly-linked list. So we get fast insertion and deletion at any position, but **no random access**.

### **Predict, then run!** Will the following code compile?
<br>

```c
////stdListBroken.cpp
// Example 7a (Predict, then run!) - can we subscript a std::list?
//%CXXFLAGS -std=c++11
#include <iostream>
#include <list>

int main(){
    std::list<int> lst = {1, 2, 3, 4};
    std::cout << lst[3] << "\n";
    return 0;
}
```

<br>

### Compile error. Internally `std::list` is a chain of nodes — there's no constant-time way to get to position 3 the way `vector`'s contiguous storage allows. The STL refuses to give us an `operator[]` that *looks* constant time but secretly walks the chain. (Notice: that's a different choice from the one we made in §4 — our `operator[]` walks behind the scenes; STL chose to expose the cost.)

### **Aside — constant time vs. linear time:**
### - **Constant time** (O(1)): the work doesn't grow with the size of the data. `vector[5]` takes one arithmetic step whether the vector has 10 or 10 million elements.
### - **Linear time** (O(n)): the work grows in proportion to the data size. Walking a linked list to position 5 takes 5 steps; to position 1000, 1000 steps.

### So `vector[i]` is constant, `linkedList[i]` is linear (specifically O(i)) — same syntax, very different cost.

### Instead, `std::list` gives us an **iterator** — a companion type that behaves like a pointer into the list. You can dereference it with `*it` to read the current element, and advance it with `++it`.
<br>

```c
////stdListIter.cpp
// Example 7b - iterators on std::list
//%CXXFLAGS -std=c++11
#include <iostream>
#include <list>

int main(){
    std::list<int> lst = {1, 2, 3, 4};

    std::list<int>::iterator it = lst.begin();   // points at first element
    std::cout << *it << "\n";                    // 1
    ++it;
    std::cout << *it << "\n";                    // 2

    return 0;
}
```

<br>

### `lst.begin()` gives an iterator at the first element. `lst.end()` gives an iterator **one past the last** element — *not* a real element, just a sentinel. Why one past? So that the same loop pattern works on every list, including the empty one — `it != lst.end()` is the natural stopping condition. Same shape as our `current != nullptr` walk in `LinkedList`. `nullptr` is the sentinel for "off the end" in our class; `lst.end()` is the sentinel for "off the end" in `std::list`.

### Here's the complete loop pattern in a runnable example. Read it carefully — you'll be writing similar loops below.
<br>

```c
////stdListLoopDemo.cpp
// Example 7c - the complete for-loop pattern using iterators (read this carefully)
//%CXXFLAGS -std=c++11
#include <iostream>
#include <list>

int main(){
    std::list<int> lst = {1, 2, 3, 4};

    for (std::list<int>::iterator it = lst.begin(); it != lst.end(); ++it){
        std::cout << *it << " ";
    }
    std::cout << "\n";    // prints: 1 2 3 4
    return 0;
}
```

# 8. Iterators as Wrapper Classes

### So what *is* an iterator, really?

### An iterator is an **object** — an instance of a small class that wraps a pointer-into-a-collection and overloads a few operators so the object can be *used* like a pointer:
### - `operator*` to read the current element
### - `operator++` to advance
### - `operator!=` (and `==`) to compare against another iterator (especially `end()`)

### For `std::list`, the wrapped pointer is essentially a `Node*`. The iterator class hides that detail and just exposes the pointer-like interface.

### Let's see this concretely by writing our own iterator for `LinkedList<T>`. Once we have it, we can write `for (auto it = list.begin(); it != list.end(); ++it)` on our own class — exactly like `std::list`.
<br>

```c
////linkedListIterator.cpp
// Example 8a - our own iterator class for LinkedList<T>
#include <iostream>
#include <string>

template <typename T>
struct Node {
    T data;
    Node<T>* next;
};

template <typename T>
class LinkedList {
private:
    Node<T>* head;

public:
    LinkedList() : head(nullptr) {}

    ~LinkedList(){
        Node<T>* current = head;
        while (current != nullptr){
            Node<T>* nxt = current->next;
            delete current;
            current = nxt;
        }
    }

    void insert(T n){
        if (head != nullptr){
            Node<T>* current = head;
            while (current->next != nullptr) current = current->next;
            Node<T>* last = new Node<T>;
            last->data = n;
            last->next = nullptr;
            current->next = last;
        } else {
            head = new Node<T>;
            head->data = n;
            head->next = nullptr;
        }
    }

    // Our iterator class - wraps a Node<T>* and exposes pointer-like operators
    class iterator {
    private:
        Node<T>* current;
    public:
        iterator(Node<T>* n) : current(n) {}

        T& operator*()  { return current->data; }              // read the element

        iterator& operator++(){                                 // advance
            current = current->next;
            return *this;
        }

        bool operator!=(const iterator& other) const {          // compare
            return current != other.current;
        }
    };

    iterator begin() { return iterator(head); }                 // first element
    iterator end()   { return iterator(nullptr); }              // one past the last
};

int main(){
    LinkedList<int> list;
    list.insert(10); list.insert(20); list.insert(30); list.insert(40);

    for (LinkedList<int>::iterator it = list.begin(); it != list.end(); ++it){
        std::cout << *it << " ";
    }
    std::cout << "\n";
    return 0;
}
```

```c
////linkedListIterator_string.cpp
// Example 8b (You try!) - print every element of LinkedList<std::string> using our iterator
#include <iostream>
#include <string>

template <typename T>
struct Node {
    T data;
    Node<T>* next;
};

template <typename T>
class LinkedList {
private:
    Node<T>* head;
public:
    LinkedList() : head(nullptr) {}
    ~LinkedList(){
        Node<T>* current = head;
        while (current != nullptr){
            Node<T>* nxt = current->next;
            delete current;
            current = nxt;
        }
    }
    void insert(T n){
        if (head != nullptr){
            Node<T>* current = head;
            while (current->next != nullptr) current = current->next;
            Node<T>* last = new Node<T>; last->data = n; last->next = nullptr;
            current->next = last;
        } else {
            head = new Node<T>; head->data = n; head->next = nullptr;
        }
    }
    class iterator {
    private:
        Node<T>* current;
    public:
        iterator(Node<T>* n) : current(n) {}
        T& operator*()  { return current->data; }
        iterator& operator++(){ current = current->next; return *this; }
        bool operator!=(const iterator& other) const { return current != other.current; }
    };
    iterator begin() { return iterator(head); }
    iterator end()   { return iterator(nullptr); }
};

int main(){
    LinkedList<std::string> names;
    names.insert("Alice"); names.insert("Bob"); names.insert("Carol");

    // your code here: print every element of names using our iterator



    std::cout << "\n";
    return 0;
}
```

### **Connecting back to `std::list`:** the iterator class for `std::list` does the same job, just more polished — it supports going backwards (`--it`), comparing iterators from different lists for safety, working with `const` containers, etc. But the shape is the same: a class that holds a pointer-into-the-data and overloads `*`, `++`, `!=`.

### One last note: when you write `auto it = list.begin();`, the compiler figures out `it`'s type for you. With long iterator types like `std::list<int>::iterator` or `LinkedList<int>::iterator`, that's a real readability win. (`auto` is filling in a long type name; it's not making the type fuzzy.)
