```c++
////activityQ4.cpp
void wastefulFunction(){
    int* pointer;
    long i = 0;
    long limit = 10;
    while(i < limit){
        if (limit < 1000000){
            ++limit;
        }
        pointer = new int;
        pointer = new int;
        delete pointer;
        pointer = new int[55000];
        pointer = new int[1500];
        pointer = new int[1000];
        ++i;
    }
    delete[] pointer;  
}
```
