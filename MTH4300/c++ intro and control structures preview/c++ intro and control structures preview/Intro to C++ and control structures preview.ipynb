### Now let's solve the mystery!

### Remember our buggy average program?

```cpp
int average = (score1 + score2) / 2;
```

### What types are involved here?

<br><br><br><br><br><br><br><br><br><br>

### `int / int` gives `int`

### 175 / 2 = 87 (not 87.5!)

### The decimal part is *truncated* (thrown away).

### How can we fix this?

<br><br><br><br><br><br><br><br><br><br>

### First attempt: just change to `double`?

### Try it. What do you get?


````python
//// averageAttempt1.cpp
// Example 3b - first attempt at fixing the average (still buggy!)
#include <iostream>
int main() {
    int score1 = 85;
    int score2 = 90;
    double average = (score1 + score2) / 2;  // changed to double
    std::cout << "Average: " << average;
    return 0;
}
````



<br><br><br><br><br><br><br><br><br><br>
### Still 87! (or 87.0)

### Why doesn't this work?

<br><br><br><br><br><br><br><br><br><br>

### The division happens *first*, then the result gets stored.

### `int / int` → `int` → then stored in `double`

### 175 / 2 = 87 → stored as 87.0

### The truncation already happened!

<br><br><br><br><br><br><br><br><br><br>

### We need to make the *division* use decimals.

### Change `2` to `2.0`:

### Try it - does it work now?


````python
//// averageFixed.cpp
// Example 3c - fixed average using 2.0
#include <iostream>
int main() {
    int score1 = 85;
    int score2 = 90;
    double average = (score1 + score2) / 2.0;  // 2.0 makes it work!
    std::cout << "Average: " << average;
    return 0;
}
````

<br><br><br><br><br><br><br><br><br><br>

### Yes! 87.5

### `int / double` → `double`

### 175 / 2.0 = 87.5

<br><br><br><br><br><br><br><br><br><br>

### Why does this work? 

When you mix types in an operation (like `int + double` or `int / double`), C++ automatically **promotes** the `int` to a `double` before doing the math. So `175 / 2.0` becomes `175.0 / 2.0`, which gives `87.5`. 

This is called **implicit type conversion** or **type promotion** — the "smaller" type gets converted to the "larger" one so no data is lost.

# 6. Preview: Control Structures

Next class, we'll work with **control structures** — loops and conditionals. You've used these in Python; here's a quick look at the C++ syntax so you're ready.

## While Loops

**Python:**
```python
i = 1
while i <= 5:
    print(i)
    i += 1
```

**C++:**
```cpp
int i = 1;
while (i <= 5) {
    std::cout << i << "\n";
    ++i;
}
```

**Key differences:**
- Parentheses `()` required around the condition
- Curly braces `{}` define the loop body (not indentation)
- `++i` is shorthand for `i = i + 1` (you can also write `i += 1` or `i++`)

*Note:* We prefer `++i` over `i++`. Both increment `i`, but `i++` makes a copy of the value before incrementing, which can be slower. For `int` the difference is often optimized away, but `++i` is better practice.

<br><br><br><br><br><br><br><br><br><br>

## Conditionals (if-else)

**Python:**
```python
if x > 0:
    print("positive")
elif x < 0:
    print("negative")
else:
    print("zero")
```

**C++:**
```cpp
if (x > 0) {
    std::cout << "positive\n";
} else if (x < 0) {
    std::cout << "negative\n";
} else {
    std::cout << "zero\n";
}
```

**Key differences:**
- Parentheses around condition
- `else if` instead of `elif`
- Curly braces for each block

<br><br><br><br><br><br><br><br><br><br>

## For Loops

**Python:**
```python
for i in range(1, 6):
    print(i)
```

**C++:**
```cpp
for (int i = 1; i <= 5; ++i) {
    std::cout << i << "\n";
}
```

**Structure:** `for (initialization; condition; update)`
