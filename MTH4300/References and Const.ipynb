# 1. Reference Variables (Review)

### Recall: a reference acts as an *alias* for an existing variable. References are declared using `&`. 

### Today, let's focus on the rules and restrictions for references, and then move on to `const`.



### A C++ reference variable must be bound to an existing variable upon declaration. It cannot be left uninitialized, and it cannot be bound to a literal value.

````c
////declaringReferenceVariables.cpp
// Example 1a
#include<iostream>
int main(){
    int& a = 1; // will this work?
    int b; // recall, it is okay to declare a variable and not initialize it right away
    int& c;
}
````

<br><br><br><br><br>
<br><br><br><br><br>

### Also, once declared, a reference variable cannot be bound to different object. 



````c
////bindingReferenceVariables.cpp
// Example 1b
#include<iostream>
int main(){
    // First, what is the the picture in memory for the following lines?
//     int a = 1;
//     int c = 2;
//     a = c;   
    
    //*******************************
    
    int a = 1;
    int& b = a; // b is now bound to the object represented by a
    int c = 2;
    b = c; // this works like it usually would, keeping in mind that "b" 
           // can be replaced by "a" in this line
    std::cout << "b = " << b << ", c = " << c << ", a = " << a << "\n"; // what prints?
    std::cout << "Addresses: &b = " << &b << ", &c = " << &c << ", &a = " << &a << "\n";
    // Note & is still used for addresses! It means something different when we put it
    // after a data type when declaring a reference variable.
}
````

<br><br><br><br><br>
<br><br><br><br><br>

### The line `b=c;` in the example above doesn't make `b` bound to `c`. Instead, this line should be read as `a = c;`. What would usually happen in this scenario? In Python, `a` and `c` would refer to the same object in memory. The same memory model is not supported in C++. A line like `a=c;` actually makes a copy of the value stored in `c` and creates a new object for `a`.

### To summarize, here are things we *cannot* do with a reference variable:
 - ### Leave it unitialized
 - ### Initialize to anything other than another *variable* of the same type
 - ### Once initialized, make it an alias for a different variable

````c
////moreReferenceVariablesPractice.cpp
// Example 1c - Which lines will print out the same value twice?
#include<iostream>
int main(){
    int x = 4;
    int& y = x;
    std::cout << x << " " << y << std::endl;
    std::cout << &x << " " << &y << std::endl; 
    
    y = 5;
    std::cout << x << " " << y << std::endl;
    
    int z = 2;
    y = z;
    std::cout << z << " " << y << std::endl;
    std::cout << &z << " " << &y << std::endl;
  
    int& w = y; // w is now also a reference to x (equivalent to int& w = x;)
    w = w + 1;
    std::cout << w << " " << y << std::endl;
    std::cout << &w << " " << &y << std::endl;
    
    int v = y;
    v = v + 1;
    std::cout << v << " " << y << std::endl;
    std::cout << &v << " " << &y << std::endl;
}
````

# 2. References as Function Parameters

### Typically, you won't see references being used in the way we have seen them in the examples above. References are particularly useful in function parameters, allowing functions to modify the original variables passed as arguments. 

````c
////referencesFunctionParameters.cpp
// Example 2a
#include<iostream>
void referenceTest(int& x){
    std::cout << &x << "\n"; //print out the address of x
}
void notReference(int x){
    std::cout << &x << "\n";
}
int main(){
    int y = 5;
    std::cout << "Address of y in main = " << &y << std::endl;
    referenceTest(y);
    notReference(y);
}
````

# 3. `const` Variables


### In C++, the ```const``` keyword is used to define constant values, meaning values that cannot be changed after they are initialized. It can be applied to variables, pointers, function arguments, member functions, and return types to ensure that the data they refer to is not modified. Let's explore its different uses.

### Const variables: When you declare a variable with const, its value cannot be changed after initialization.

````c
////constVariables.cpp
// Example 3a
int main(){
    const int maxValue = 100;
    // maxValue = 50; // Error: cannot modify a const variable
}
````


<br><br><br><br><br>
<br><br><br><br><br>

### There are different variations of using `const` with pointers, and each has a distinct meaning. The position of `const` relative to the `*` matters.

### The most common and practical one is a **Pointer to a Const Value** (`const int*`). This means that the value the pointer points to cannot be modified through this pointer, but the pointer itself can change to point to something else. You'll encounter this in C code and older C++ code, since C doesn't have references — `const int*` was the only way to pass data without copying while promising not to modify it.

````c
////constPointers.cpp
// Example 3b - Pointer to a const value
int main(){
    const int* ptr;
    int x = 10;
    int y = 20;
    ptr = &x;  // Okay
    *ptr = 15; // Error: Cannot modify the value of x through ptr
    ptr = &y;  // Okay
}
````

<br><br><br><br><br>
<br><br><br><br><br>

### For completeness: if you put `const` **after** the `*`, it makes the **pointer itself** const — it can't be reassigned to point somewhere else, but the value it points to can still be changed. You can also combine both. These are less common, but you should know they exist.

### A helpful trick: **read the declaration right to left**.
- ### `const int* ptr` - "ptr is a pointer to an int that is const" - can't change the **value**
- ### `int* const ptr` - "ptr is a const pointer to an int" - can't change the **pointer**
- ### `const int* const ptr` - "ptr is a const pointer to an int that is const" - can't change **either**

````c
////constPointers2and3.cpp
// Example 3c - Const pointer and const pointer to const value
int main(){
    int x = 10;
    int y = 20;
    
    // Const pointer: pointer can't change, but value can
    int* const ptr1 = &x;
    *ptr1 = 15; // Okay: Can modify the value of x through ptr1
    //ptr1 = &y;  // Error: ptr1 itself is const and cannot be reassigned
    
    // Const pointer to const value: neither can change
    const int* const ptr2 = &x;
    //*ptr2 = 15; // Error: Cannot modify the value through ptr2
    //ptr2 = &y;  // Error: ptr2 itself is const and cannot be reassigned
}
````

### Just as `const int*` prevents modification through a pointer, `const int&` prevents modification through a reference. A `const` reference is declared using the `const` keyword before the type and the reference operator `&`. It creates a reference that cannot be used to modify the original variable.

````c
////constReference.cpp
// Example 3e
#include<iostream>
int main(){
    int x = 1;
    const int& y = x;
   //y = y + 1;
    x = x + 1;
    
}
````

<br><br><br><br><br>
<br><br><br><br><br>

### Important: declaring a `const` pointer or `const` reference to a variable does **not** make the original variable `const`. It only restricts modification *through that particular alias*. In the example above, we can't modify `x` through `y`, but we can still modify `x` directly. The same applies to `const int* ptr = &x` — we can't do `*ptr = 5`, but we can still do `x = 5`.

# 4.`const` Function Arguments and `const` Objects

### Declaring function parameters as `const` ensures that they are not modified within the function. This is useful for passing parameters by reference. The function is making a promise not to change the object passed as a parameter, even though it is passed by reference.

````c
////constArguments.cpp
// Example 4a
#include<iostream>
void printValue(const int& value) {
    // value = 10; // Error: Cannot modify value since it is const
    std::cout << value << std::endl;
}
int main(){
    int x = 1;
    printValue(x); // signature line of function is saying const int& value = x;
    printValue(3); // 3 is just 3...what would it mean to change 3?
    
}
````


### Passing a literal argument to a function that accepts a `const` argument is valid. The `const` keyword signifies that the function will not modify the argument passed to it. A literal value is inherently immutable, so it aligns with the function's promise.


<br><br><br><br><br>
<br><br><br><br><br>

````c
////constArguments2.cpp
// Example 4b
#include<iostream>
void printFirstLetter(const std::string& book){
    std::cout << book[0] << "\n";
}

int main(){
    std::string myBook = "Imagine this string is the full text of a long novel and not just one sentence.";
    printFirstLetter(myBook);
}
````

### When an object is declared as `const`, you cannot modify its data members or call any non-`const` member functions on that object.

````c
////constObjects.cpp
// Example 4e
#include<iostream>
class Example {
private:
    int value;

public:
    Example(int v){
        this->value = v;
    }

    int getValue() const {  // Const member function
        return this->value;
    }

    void setValue(int v) {  // Non-const member function
        this->value = v;
    }
};

int main(){
    const Example obj(5);
    obj.getValue();    // Okay: getValue() is a const member function
    obj.setValue(10);  // Error: Cannot call non-const member function on a const object
}
````

# 5. Circling Back: Why Does the Copy Constructor Use `const&`?

### Now that we understand both references and `const`, let's revisit the copy constructor signature:

### `Character(const Character& other)`

### There are two things going on here. Why `&` (reference)? And why `const`?

### **Why `&` (reference)?**

### Think about what would happen if the copy constructor took its argument by value instead of by reference:

### `Character(Character other)  // pass by value — NO!`

### To pass by value, C++ needs to *copy* the argument into the parameter. But how does it copy an object? By calling the copy constructor! So the copy constructor would call itself to copy `other`, which would call itself again, and again... **infinite recursion**.

### Using a reference (`&`) avoids this — no copy is made, `other` is just an alias for the original object.

<br><br><br><br><br>
<br><br><br><br><br>

### **Why `const`?**

### The copy constructor's job is to read from the original object and create a new copy. It should **never modify** the original. Adding `const` makes that promise explicit — the compiler will stop us if we accidentally try to change `other` inside the copy constructor.

### It also lets us copy from `const` objects. Without `const`, we couldn't do this:



````c
////whyConstRef.cpp
// Example 5a - Why the copy constructor needs const&
#include<iostream>
#include<string>

class Character{
public:
    std::string name_;
    
    Character(std::string n){
        this->name_ = n; 
    }
    
    // What if we wrote the copy constructor WITHOUT const?
    // Character(Character& other){  // no const!
    //     this->name_ = other.name_;
    // }
    
    // With const:
    Character(const Character& other){
        this->name_ = other.name_;
    }
    
    void display() const {
        std::cout << this->name_ << "\n";
    }
};

void printCharacter(const Character& c){
    Character copy = c;  // this calls the copy constructor
    // If copy constructor were Character(Character& other),
    // this would FAIL — c is const, but other is non-const reference!
    copy.display();
}

int main(){
    Character mario("Mario");
    printCharacter(mario);
}
````

<br><br><br><br><br>
<br><br><br><br><br>

### To summarize: `const&` in the copy constructor gives us the best of both worlds:
- ### `&` — avoids infinite recursion (no copy needed to pass the argument)
- ### `const` — promises not to modify the original, and allows copying from `const` objects
