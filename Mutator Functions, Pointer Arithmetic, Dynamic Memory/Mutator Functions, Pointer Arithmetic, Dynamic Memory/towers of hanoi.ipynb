### The puzzle:

### There are 3 pegs and *n* rings of different sizes, all starting on the first peg (largest on the bottom, smallest on top). The goal is to move the entire stack from peg 1 to peg 3, following these rules:

### 1. Only one ring may be moved at a time.
### 2. Each move takes the top ring from one peg and places it on another peg.
### 3. No ring may be placed on top of a smaller ring.

### We want to write a recursive function that prints out the step-by-step instructions for solving the puzzle.

### Think about these questions:

### 1. What is the base case? (What's the simplest version of this problem that you can solve directly?)

### 2. For *n* > 1, how can you express the problem of moving *n* rings in terms of moving *n* − 1 rings?

### 3. Key observation: when moving the *n* − 1 smaller rings around, can you completely ignore the largest ring? Why or why not?
