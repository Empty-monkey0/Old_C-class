# 1. Mutator Functions

### In the previous notebook (Examples 4a and 4b), we tried to write a function `addTwoToXAndThreeToY` that would modify two variables in `main`. It didn't work.

### Why not?

<br><br><br><br><br><br><br><br><br><br>

### The function received *copies* of the values, stored at different addresses. The originals were never touched.

### What would the function need to receive instead, so that it could actually modify the originals?

<br><br><br><br><br><br><br><br><br><br>

### In terms of our mailbox analogy: instead of photocopying the *contents* of each mailbox and sending those to the function, we send copies of the *keys*. The function uses the keys to open the original mailboxes and modify what's inside.

### So what should this function do?
* ### Take the **keys to** (addresses of) `x` and `y` as parameters
* ### Use the key to open `x`'s mailbox, read the value, add 2, and store it back
* ### Use the key to open `y`'s mailbox, read the value, add 3, and store it back

<br><br><br><br><br><br><br><br><br><br>

### Translating to C++:
* ### "key to `x`" / "key to `y`" are pointers: `int* aX` and `int* aY`
* ### "open the mailbox" is a dereference: `*aX`
* ### In `main`, we pass the keys using `&`: `addTwoToXAndThreeToY(&x, &y)`

### The type before the `*` indicates what the pointer points to. `int* aX` means "`aX` is a pointer to an `int`".

### What would we write if we were pointing to a `double`?

<br><br><br><br><br><br><br><br><br><br>

````c
//// mutators.cpp
// Example 1a
#include <iostream>
void addTwoToXAndThreeToY(int* aX, int* aY){
    *aX = *aX + 2;
    *aY = *aY + 3;
}
int main(){
    int x, y;
    x = 7; y = 17;
    std::cout << "old x = " << x << "\n";
    std::cout << "old y = " << y << "\n";
    addTwoToXAndThreeToY(&x, &y);
    std::cout << "new x = " << x << "\n";
    std::cout << "new y = " << y << "\n";
}
````

# 2. Pointer Arithmetic

### Arrays are stored as a contiguous block of memory, meaning the elements are one after another, with no gaps.

### What happens when we use the array's name in an expression, for example, when we print it?

<br><br><br><br><br><br><br><br><br><br>

````c
//// arrayAddresses.cpp
// Example 2a
#include <iostream>
int main(){
    int arr[] = {10, 20, 30};
    std::cout << "arr:      " << long(arr) << "\n";
    std::cout << "&arr:     " << long(&arr) << "\n";
    std::cout << "&arr[0]:  " << long(&arr[0]) << "\n";
}
````


<br><br><br><br><br><br><br><br><br><br>

### Since `arr` evaluates to a pointer, we can do arithmetic with it.

````c
//// pointerArithInt.cpp
// Example 2b
#include <iostream>
int main(){
    int arr[] = {10, 20, 30};
    int* ptr = arr;  // ptr points to arr[0]
    std::cout << "ptr:   " << long(ptr) << "  *ptr:   " << *ptr << "\n";
    ++ptr;           // move to the next element
    std::cout << "ptr:   " << long(ptr) << "  *ptr:   " << *ptr << "\n";
    ++ptr;           // move to the next element
    std::cout << "ptr:   " << long(ptr) << "  *ptr:   " << *ptr << "\n";
}
````

# 3. Dynamic Memory

### So far, every variable we've created lives on the **stack** — its lifetime is tied to the scope it was declared in. When the scope ends, the variable's memory is automatically freed.

### What happens here?

````c
//// danglingMotivation.cpp
// Example 3a
#include <iostream>
int* createValue(){
    int x = 3;
    std::cout << "x = " << x << ", address: " << long(&x) << "\n";
    return &x;  // returning the address of a local variable...
}
int main(){
    int* ptr = createValue();
    std::cout << "ptr holds address: " << long(ptr) << "\n";
}
````

<br><br><br><br><br><br><br><br><br><br>

### The `new` operator allocates memory on the **heap** — a separate region of memory from the stack. Unlike stack memory, heap memory is not tied to any scope. It persists until the programmer explicitly frees it using `delete`.

### `new` does the following:
* ### Allocates space for an object of the given type on the heap
* ### Initializes the object (if an initial value is provided)
* ### Returns the *address* of the newly created object

### Since `new` returns an address, we store the result in a pointer.

<br><br><br><br><br><br><br><br><br><br>

````c
//// newAndDelete.cpp
// Example 3b
#include <iostream>
int main(){
    int* ptr = new int(3);    // allocate an int on the heap, initialize to 3
    std::cout << "Value: " << *ptr << "\n";
    std::cout << "Address: " << long(ptr) << "\n";
    
    int* ptr2 = new int;      // allocate without initializing
    *ptr2 = 35;               // assign a value afterwards
    std::cout << "Value: " << *ptr2 << "\n";
    
    delete ptr;   // free the memory
    delete ptr2;  // free the memory
    ptr = nullptr;   // good practice: avoid dangling pointer
    ptr2 = nullptr;
}
````

### What do you notice about the address of the heap variable compared to typical stack addresses?

<br><br><br><br><br><br><br><br><br><br>

````c
//// stackVsHeapAddresses.cpp
// Example 3c
#include <iostream>
int main(){
    int stackVar = 10;
    int* heapVar = new int(20);
    std::cout << "Stack variable address: " << long(&stackVar) << "\n";
    std::cout << "Heap variable address:  " << long(heapVar) << "\n";
    std::cout << "Pointer itself (on stack): " << long(&heapVar) << "\n";
    delete heapVar;
    heapVar = nullptr;
}
````

### We can also allocate arrays on the heap using `new[]`. What address range do you expect the elements to be in?

````c
//// heapArrays.cpp
// Example 3f
#include <iostream>
int main(){
    int* arr = new int[5];  // allocate an array of 5 ints on the heap
    for (int i = 0; i < 5; i++){
        arr[i] = i * 2;
        *(arr + i) = i * 2; //equivalent to previous line
        std::cout << "arr[" << i << "] = " << arr[i]
                  << "  address: " << long(&arr[i]) << "\n";
    }
    delete[] arr;  // free the array — note the []
    arr = nullptr;
}
````

### Important: `new` pairs with `delete`, and `new[]` pairs with `delete[]`. Using the wrong form is undefined behavior.

<br><br><br><br><br><br><br><br><br><br>

### What's wrong with this code?

````c
//// memoryLeak.cpp
// Example 4b
#include <iostream>
int main(){
    int* ptr = new int(5);
    delete ptr; // free mem for 5 before overwrite ptr with 
    ptr = new int(10);
    
    std::cout << *ptr << "\n";
    delete ptr;
    ptr = nullptr;
}
````

<br><br><br><br><br><br><br><br><br><br>

### Key rules for working with dynamic memory:
* ### Always `delete` what you `new` (and `delete[]` what you `new[]`)
* ### Set pointers to `nullptr` after `delete` to avoid dangling pointers
* ### Initialize pointers to `nullptr` if you don't have a valid address yet
* ### Don't reassign a pointer to heap memory without freeing the old allocation first
* ### When possible, prefer `std::vector` over raw heap arrays

<br><br><br><br><br><br><br><br><br><br>

### Modern C++ provides **smart pointers** that handle deallocation automatically. For example, `std::unique_ptr` frees the heap memory when the pointer goes out of scope:

````c
//// smartPointers.cpp
//% CXXFLAGS -std=c++14
// Example 4c
#include <memory>
#include <iostream>
int main(){
    std::unique_ptr<int> ptr = std::make_unique<int>(10);
    std::cout << "Value: " << *ptr << "\n";
    // no delete needed — automatically freed when ptr goes out of scope
}
````
